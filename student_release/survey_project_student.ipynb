## 2. Confirm the required functions are available


```python
required_functions = [
    "readsurvey",
    "listentries",
    "victory",
    "plurality",
    "borda",
    "condorcet",
    "irv",
    "approval",
]

for function_name in required_functions:
    available = hasattr(blackbox, function_name)
    print(f"{function_name}: {available}")
```

## 3. Quick functionality test

This checks one simple helper function before working with a full survey file.


```python
test_rankings = ["1", "", "3", "2"]
converted = blackbox.listentries(test_rankings)

print("Original:", test_rankings)
print("Converted:", converted)

assert converted == [1, 0, 3, 2]
print("Basic test passed.")
```

## 4. Read your survey CSV file

Change `survey.csv` below if your instructor provided a different filename.


```python
survey_file = "survey.csv"

names, votes = blackbox.readsurvey(survey_file)

print("Candidates:")
print(names)

print("\nNumber of survey responses:", len(votes))

print("\nFirst few responses:")
for vote in votes[:5]:
    print(vote)
```

## 5. Plurality voting


```python
plurality_results, plurality_winner, plurality_max = blackbox.plurality(
    names.copy(),
    [vote.copy() for vote in votes]
)

print("Plurality results:", plurality_results)
print("Winner:", plurality_winner)
print("Top vote count:", plurality_max)
```

## 6. Borda count


```python
borda_results, borda_winner, borda_max = blackbox.borda(
    names.copy(),
    [vote.copy() for vote in votes]
)

print("Borda results:", borda_results)
print("Winner:", borda_winner)
print("Top score:", borda_max)
```

## 7. Condorcet method


```python
condorcet_results, condorcet_winner, condorcet_max = blackbox.condorcet(
    names.copy(),
    [vote.copy() for vote in votes]
)

print("Condorcet winner:", condorcet_winner)
print("Maximum pairwise wins:", condorcet_max)
```

## 8. Approval voting


```python
approval_results, approval_winner, approval_max = blackbox.approval(
    names.copy(),
    [vote.copy() for vote in votes]
)

print("Approval results:", approval_results)
print("Winner:", approval_winner)
print("Top approval count:", approval_max)
```

## 9. Instant-runoff voting (IRV)

The IRV function modifies the candidate and vote lists while it runs, so this notebook passes copies of the original data.


```python
import copy

irv_names = names.copy()
irv_votes = copy.deepcopy(votes)

irv_results, irv_winner, irv_max, irv_rounds = blackbox.irv(
    irv_names,
    irv_votes
)

print("IRV results:", irv_results)
print("Winner:", irv_winner)
print("Top vote count:", irv_max)
print("Rounds:", irv_rounds)
```
